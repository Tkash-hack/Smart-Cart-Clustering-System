```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("smartcart_customers.csv")
```

```python
df.isnull().sum()
df.head()
```

# DATA PREPROCESSING

```python
# 1. HANDLE MISSING VALUES
df["Income"]=df["Income"].fillna(df["Income"].median())
```

# Feature Engineering

```python
# 1. Age
df["Age"]=2026-df["Year_Birth"]

# 2. Dt_Customer
df["Dt_Customer"]=pd.to_datetime(df["Dt_Customer"],dayfirst=True)
Reference=df["Dt_Customer"].max()
df["Customer_tenure_days"]=(Reference-df["Dt_Customer"]).dt.days
```

```python
# 3. Num of children
df["Children"]=df["Kidhome"]+df["Teenhome"]

# 4. Total Spending
df["Total Spending"]=df["MntWines"]+df["MntFruits"]+df["MntMeatProducts"]+df["MntFishProducts"]+df["MntSweetProducts"]+df["MntGoldProds"]

# 5. Education
df["Education"]=df["Education"].replace({
    "PhD":"Postgraduate",
    "Master":"Postgraduate",
    "2n Cycle":"Undergraduate",
    "Basic":"Undergraduate",
    "Graduation":"Graduate"
})
df["Education"].value_counts()

# 6. Marital Status
df["Living_Status"]=df["Marital_Status"].replace({
    "Married" : "Partner",
    "Together":"Partner",
    "Single":"Alone",
    "Divorced":"Alone",
    "Widow":"Alone",
    "Absurd":"Alone",
    "YOLO":"Alone",
    
})
```

```python
df.columns
df["Marital_Status"].value_counts()
```

# Drop Columns

```python
cols=["ID","Year_Birth","Marital_Status","Kidhome","Teenhome","Dt_Customer"]
spending_cols=["MntWines","MntFruits","MntMeatProducts","MntFishProducts","MntGoldProds","MntSweetProducts"]
cols_to_drop=cols+spending_cols
df_cleaned=df.drop(cols_to_drop,axis=1)
```
